# IT Services Performance Analytics - Exploratory Analysis

This notebook analyzes synthetic generated data for a fictional IT services company.

The analysis includes service contribution, project completion, delays, budget variances, client satisfaction, and team performance indicators.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')

clean_dir = '../data/cleaned'
clients = pd.read_csv(f'{clean_dir}/clients_cleaned.csv')
projects = pd.read_csv(f'{clean_dir}/projects_cleaned.csv')
team = pd.read_csv(f'{clean_dir}/team_performance_cleaned.csv')
quality = pd.read_csv(f'{clean_dir}/project_quality_cleaned.csv')
finance = pd.read_csv(f'{clean_dir}/monthly_finance_cleaned.csv')

print(clients.shape, projects.shape, team.shape, quality.shape, finance.shape)
```

```python
# Project distribution by service
service_counts = projects['service'].value_counts()
plt.figure(figsize=(8, 5))
service_counts.plot(kind='bar', color='steelblue')
plt.title('Project Distribution by Service')
plt.xlabel('Service')
plt.ylabel('Projects')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Revenue by service
finance_grouped = finance.groupby('service')['revenue'].sum().sort_values()
plt.figure(figsize=(8, 5))
finance_grouped.plot(kind='barh', color='darkgreen')
plt.title('Revenue by Service')
plt.xlabel('Revenue')
plt.ylabel('Service')
plt.tight_layout()
plt.show()
```

```python
# Cost by service
cost_grouped = finance.groupby('service')['operating_cost'].sum().sort_values()
plt.figure(figsize=(8, 5))
cost_grouped.plot(kind='barh', color='firebrick')
plt.title('Operating Cost by Service')
plt.xlabel('Cost')
plt.ylabel('Service')
plt.tight_layout()
plt.show()
```

```python
# Profit by service
profit_grouped = finance.groupby('service')['profit'].sum().sort_values()
plt.figure(figsize=(8, 5))
profit_grouped.plot(kind='barh', color='darkorange')
plt.title('Profit by Service')
plt.xlabel('Profit')
plt.ylabel('Service')
plt.tight_layout()
plt.show()
```

```python
# Monthly revenue trend
finance['month'] = pd.to_datetime(finance['month'])
monthly_revenue = finance.groupby('month')['revenue'].sum()
plt.figure(figsize=(10, 5))
monthly_revenue.plot(kind='line', color='blue')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Monthly profit trend
monthly_profit = finance.groupby('month')['profit'].sum()
plt.figure(figsize=(10, 5))
monthly_profit.plot(kind='line', color='green')
plt.title('Monthly Profit Trend')
plt.xlabel('Month')
plt.ylabel('Profit')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Project completion status
status_counts = projects['status'].value_counts()
plt.figure(figsize=(6, 5))
status_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Projects by Status')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
# Delay distribution
projects['delay_days'] = (pd.to_datetime(projects['actual_end_date']) - pd.to_datetime(projects['planned_end_date'])).dt.days
plt.figure(figsize=(8,5))
sns.histplot(projects['delay_days'], bins=20, kde=False)
plt.title('Delay Distribution')
plt.xlabel('Delay Days')
plt.ylabel('Projects')
plt.tight_layout()
plt.show()
```

```python
# Average delay by service
projects['delay_days'] = (pd.to_datetime(projects['actual_end_date']) - pd.to_datetime(projects['planned_end_date'])).dt.days
avg_delay_service = projects.groupby('service')['delay_days'].mean().sort_values(ascending=False)
plt.figure(figsize=(8,5))
avg_delay_service.plot(kind='barh', color='purple')
plt.title('Average Delay by Service')
plt.xlabel('Average Delay Days')
plt.ylabel('Service')
plt.tight_layout()
plt.show()
```

```python
# Budget overrun
overrun_pct = (projects['actual_cost'] > projects['budget']).mean() * 100
print(f'Budget overrun percentage: {overrun_pct:.2f}%')
budget_overrun = (projects['actual_cost'] > projects['budget']).sum()
plt.figure(figsize=(6,4))
plt.bar(['Within Budget', 'Over Budget'], [len(projects)-budget_overrun, budget_overrun], color=['green','red'])
plt.title('Budget Overrun Count')
plt.ylabel('Projects')
plt.show()
```

```python
# Top clients by revenue
client_projects = projects.merge(clients, on='client_id', how='left')
client_revenue = client_projects.groupby('client_name')['budget'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(8,5))
client_revenue.plot(kind='barh', color='darkslateblue')
plt.title('Top Clients by Revenue')
plt.xlabel('Revenue')
plt.ylabel('Client')
plt.tight_layout()
plt.show()
```

```python
# Team utilization and productivity relation
team_project = team.merge(projects, on='project_id', how='left')
plt.figure(figsize=(8,5))
sns.scatterplot(data=team_project, x='utilization_rate', y='productivity_score', hue='service')
plt.title('Productivity vs Utilization')
plt.xlabel('Utilization Rate %')
plt.ylabel('Productivity Score')
plt.tight_layout()
plt.show()
```
